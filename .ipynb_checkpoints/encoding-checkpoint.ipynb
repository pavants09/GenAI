```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained('bert-base-cased')
```

```python
encoding_data = tokenizer(['Hello World','im pavan gowda'])
print(encoding_data)
```

# Decode that embedding input id to orginal data

```python
tokenizer.decode(encoding_data['input_ids'])
```

# Encoding text

### return_tensors

```python
encoding_data = tokenizer('Hello World','im pavan gowda',return_tensors="pt")
print(encoding_data)
```

### Padding inputs


```python
encoding_data = tokenizer(["Hello World","i'm pavan gowda"], padding = True, return_tensors = "pt")
print(encoding_data)
```

### Truncating inputs

```python
encoding_data = tokenizer(
    ["Hello World","i'm pavan gowda"],
    padding=True,
    truncation=True,
    max_length=5,
    return_tensors="pt",
)
print(encoding_data)
```
