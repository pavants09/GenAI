```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
```

# This Below steps displays how inside tokenizers work 

```python
tokens = tokenizer.tokenize("Hello World!")
tokens
```

```python
inputIds = tokenizer.convert_tokens_to_ids(tokens)
inputIds
```

# This is Direct Convert Raw Text to Tokens

```python
directConvertToToken = tokenizer("Hello World!")
directConvertToToken
```
